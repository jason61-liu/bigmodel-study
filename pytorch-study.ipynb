```python
a = [1, 2, 3]
b = [4, 5, 6]

# ❌ list 不支持直接相加
c = a + b
print(c)  # 输出: [1, 2, 3, 4, 5, 6] （拼接，不是逐元素相加）

# ✅ numpy 支持逐元素加法
import numpy as np
a_np = np.array([1, 2, 3])
b_np = np.array([4, 5, 6])
c_np = a_np + b_np
print(c_np)  # 输出: [5 7 9]
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib



# 设置中文字体和解决负号显示问题
matplotlib.rcParams['font.sans-serif'] = ['.Aqua Kana']  # 用黑体显示中文
matplotlib.rcParams['axes.unicode_minus'] = False  # 解决负号 '-' 显示为方块的问题

# 生成10000个标准高斯分布的样本
samples = np.random.randn(10000)

# 绘制直方图
plt.hist(samples, bins=100, density=True, alpha=0.6, color='g')

# 绘制理论上的标准高斯分布PDF
x = np.linspace(-4, 4, 100)
pdf = (1 / np.sqrt(2 * np.pi)) * np.exp(-x**2 / 2)
plt.plot(x, pdf, 'k', linewidth=2)

plt.title('标准高斯分布')
plt.show()
```
