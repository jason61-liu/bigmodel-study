```python
import torch
import torch.nn as nn

# 创建一个 ReLU 层
relu = nn.ReLU()

# 创建一个包含负值和正值的随机张量作为输入
input_tensor = torch.randn((2, 3))

# 应用 ReLU 函数
output_tensor = relu(input_tensor)

print("Input tensor:\n", input_tensor)
print("Output tensor after applying ReLU:\n", output_tensor)
```
